```python
import pandas as pd

# Load dataset (make sure the file is uploaded)
df = pd.read_csv("large-762M-k40.test.csv")

# Display first few rows and column names
print("Columns:", df.columns)
print(df.head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Assume 'text' column contains the GPT-2 answers
answers = df['text'].dropna().tolist()

# Vectorize answers
vectorizer = TfidfVectorizer()
answer_vectors = vectorizer.fit_transform(answers)
```

```python
def get_best_answer(user_input):
    user_vec = vectorizer.transform([user_input])
    similarities = cosine_similarity(user_vec, answer_vectors)
    best_idx = similarities.argmax()
    return answers[best_idx]
```

```python
print("Ask me anything I'll try to answer! Type 'exit' to stop.\n")

while True:
    user_input = input("You: ")
    if user_input.lower() in ['exit', 'quit']:
        print("Bot: Goodbye! 🌴")
        break

    answer = get_best_answer(user_input)

    # Limit answer to the first 1000 words
    word_limit = 100
    words = answer.split()

    if len(words) > word_limit:
        answer = ' '.join(words[:word_limit]) + '... [ect.]'

    print("Bot:", answer)
```
